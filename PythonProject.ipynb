```python
df = df.drop(columns=['Unnamed: 33', 'Unnamed: 34'])
df.shape
```

```python
print("Duplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

```python
df.columns = df.columns.str.strip().str.replace(' ', '_')
df.columns
```

```python
df['Return_Reason'] = df['Return_Reason'].replace('—', 'Not Returned')
df['Return_Reason'].unique()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%y')
df['Shipping_Date'] = pd.to_datetime(df['Shipping_Date'], format='%d-%m-%y')
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], format='%d-%m-%y')

df[['Order_Date','Shipping_Date','Delivery_Date']].info()
```

```python
print("Category:", df['Category'].unique())
print("Sub_Category:", df['Sub_Category'].unique())
print("Brand:", df['Brand'].unique())
print("Region:", df['Region'].unique())
print("Payment_Method:", df['Payment_Method'].unique())
print("Order_Status:", df['Order_Status'].unique())
print("Customer_Segment:", df['Customer_Segment'].unique())
```

```python
for col in ['Category','Sub_Category','Brand','Region','Payment_Method','Order_Status','Customer_Segment']:
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
high_value = df[df['Sales_Amount'] > 10000]
print(high_value.shape[0], "orders with Sales > 10,000")
high_value.head()
```

```python
returned_electronics = df[(df['Category']=='Electronics') & (df['Order_Status']=='Returned')]
print(returned_electronics.shape[0], "returned Electronics orders")
```

```python
numeric_cols = ['Age','Quantity','Unit_Price','Discount_Percent','Sales_Amount','Profit_Amount','Profit_Margin','Rating']
correlation = df[numeric_cols].corr()
correlation
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.groupby('Category')['Sales_Amount'].sum().sort_values(ascending=False).plot(kind='bar', color='steelblue')
plt.title("Category-wise Total Sales")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%y')

monthly = df.groupby(df['Order_Date'].dt.to_period('M'))['Sales_Amount'].sum()

plt.figure(figsize=(10,5))
monthly.plot(kind='line', marker='o', color='green')
plt.title("Monthly Sales Trend")
plt.ylabel("Sales Amount")
plt.xlabel("Month")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['Sales_Amount'].plot(kind='hist', bins=50, color='orange')
plt.title("Distribution of Sales Amount")
plt.xlabel("Sales Amount")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,7))
df['Region'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Orders Share by Region")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
numeric_cols = ['Age','Quantity','Unit_Price','Discount_Percent','Sales_Amount','Profit_Amount','Profit_Margin','Rating']

plt.figure(figsize=(9,6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Category', y='Sales_Amount', data=df)
plt.title("Sales Amount Outliers by Category")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Insight 1: Best performing month
monthly_sales = df.groupby(df['Order_Date'].dt.month_name())['Sales_Amount'].sum().sort_values(ascending=False)
print("Best month:", monthly_sales.index[0], "-", round(monthly_sales.iloc[0]))
```

```python
# Insight 2: Best category
category_sales = df.groupby('Category')['Sales_Amount'].sum().sort_values(ascending=False)
print("Top category:", category_sales.index[0])
```

```python
# Insight 3: Return rate
return_rate = df['Return_Flag'].value_counts(normalize=True) * 100
print("Return rate:", round(return_rate['Yes'], 2), "%")
```

```python
# Insight 4: Payment method
print("Most used payment method:", df['Payment_Method'].value_counts().index[0])
```

```python
# Insight 5: Online vs Store
channel = df.groupby('Sales_Channel')['Sales_Amount'].sum()
print("Online sales:", round(channel['Online']), "| Store sales:", round(channel['Store']))
```
